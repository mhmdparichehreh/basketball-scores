Data Augmentation:

```python
import pandas as pd

# 2. Define the exact column swaps
swap_mapping = {
    "team1q1": "team2q1", "team2q1": "team1q1",
    "team1q2": "team2q2", "team2q2": "team1q2",
    "team1q3": "team2q3", "team2q3": "team1q3",
    "team1q4": "team2q4", "team2q4": "team1q4",
    "team1tot": "team2tot", "team2tot": "team1tot"
}

# 3. Create the swapped copy
# Renaming the columns means "team1q1" data is now labeled "team2q1"
df_swapped = df.rename(columns=swap_mapping)

# 4. Append the swapped data to the original dataset
df_augmented = pd.concat([df, df_swapped], ignore_index=True)

print("Original Data Size:", len(df))
print("Augmented Data Size:", len(df_augmented))
print("\nAugmented DataFrame:")
df_augmented
```

K-Means implementation:

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler

# 2. Preprocessing
# Convert categorical 'gender' to numerical so KNN can calculate distance
df_augmented['gender_num'] = df_augmented['gender'].map({'M': 0, 'F': 1})

# Define our feature matrix
features = ["gender_num", "team1q1", "team2q1", "team1q2", "team2q2", "team1q3", "team2q3"]
X = df_augmented[features]

# Scale the features (Crucial for KNN!)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Fit the KNN Model
# n_neighbors=5 as requested. 'euclidean' is the standard straight-line distance.
knn = NearestNeighbors(n_neighbors=5, metric='euclidean')
knn.fit(X_scaled)

# 4. Create the search function
def find_similar_matches(input_gender, t1q1, t2q1, t1q2, t2q2, t1q3, t2q3):
    """Finds the 5 matches closest to the provided input."""

    # Encode input gender to match our training data
    gen_num = 1 if input_gender.upper() == 'F' else 0

    # Format as a 2D array and scale it using the SAME scaler
    input_data = pd.DataFrame(
        [[gen_num, t1q1, t2q1, t1q2, t2q2, t1q3, t2q3]],
        columns=features
    )
    input_scaled = scaler.transform(input_data)

    # Calculate distances and get the indices of the closest matches
    distances, indices = knn.kneighbors(input_scaled)

    # Fetch the actual rows from the original dataframe
    # We use indices[0] because kneighbors returns an array of arrays
    closest_matches = df_augmented.iloc[indices[0]].copy()

    # Optional: Add the distance score so you can see how close they are
    closest_matches['match_distance'] = distances[0]

    # Drop the temporary numerical gender column for cleaner output
    return closest_matches

# --- Example Usage ---
# Let's search for a Men's game with roughly these quarter scores:
results = find_similar_matches('M', 24, 21, 23, 25, 27, 24)

print("Top 5 Closest Matches:\n")
print(results)
```

```python
import pandas as pd
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler

# 2. Feature Engineering: Calculate the combined quarter sums
df['q1_sum'] = df['team1q1'] + df['team2q1']
df['q2_sum'] = df['team1q2'] + df['team2q2']
df['q3_sum'] = df['team1q3'] + df['team2q3']
df['q4_sum'] = df['team1q4'] + df['team2q4']
df['tot_dif'] = (df['team1q1'] + df['team1q2'] + df['team1q3'] - df['team2q1'] - df['team2q2'] - df['team2q3']).abs() #difference between two teams until 3rd quarter

# Convert gender to numerical
df['gender_num'] = df['gender'].map({'M': 0, 'F': 1})

# 3. Define our new feature matrix
features = ["gender_num", "q1_sum", "q2_sum", "q3_sum", "tot_dif"]
X = df[features]

# 4. Scale and Fit the KNN Model
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

knn = NearestNeighbors(n_neighbors=10, metric='euclidean')
knn.fit(X_scaled)

# 5. Create the search function for summed scores
def find_similar_by_quarter_totals(input_gender, q1_total, q2_total, q3_total, dif):
    """Finds the 5 matches with the most similar combined quarter scores."""

    # Encode input gender
    gen_num = 1 if input_gender.upper() == 'F' else 0

    # Format and scale input
    input_data = pd.DataFrame(
        [[gen_num, q1_total, q2_total, q3_total, dif]],
        columns=features
    )
    input_scaled = scaler.transform(input_data)

    # Get neighbors
    distances, indices = knn.kneighbors(input_scaled)

    # Fetch the original rows
    closest_matches = df.iloc[indices[0]].copy()
    closest_matches['match_distance'] = distances[0]

    # Return just the relevant columns for a cleaner view
    return closest_matches

# --- Example Usage ---
# Searching for a Men's game where the combined quarter totals were 45, 46, and 50
results = find_similar_by_quarter_totals('M', 45, 46, 50, 5)

print("Top 5 Matches by Combined Quarter Scores:\n")
print(results)
```

```python
gen = 'F'
q1, q2, q3 = 32, 28, 26
dif = 12
results = find_similar_by_quarter_totals(gen, q1, q2, q3, dif)

print("Top 5 Matches by Combined Quarter Scores:\n")
print(results)
print("Min: ", results['q4_sum'].min(), ", Max: ", results['q4_sum'].max())
print("(", q1+q2+q3+results['q4_sum'].min(), ",", q1+q2+q3+results['q4_sum'].max(),")")
```
